```python
import pandas as pd
import gradio as gr
import pickle
import statsmodels.api as sm
```

```python
with open("sales_prediction_model.pkl", "rb") as f:
    model = pickle.load(f)
```

```python
def predict_sales(tv, radio, newspaper):
    input_data = pd.DataFrame({
        'TV': [tv],
        'Radio': [radio],
        'Newspaper': [newspaper]
    })
    prediction = model.predict(input_data)[0]
    return f"📈 Estimated Sales: {prediction:.2f} units"
```

```python
title = "📊 Advertisement Budget → Sales Prediction"
description = """
Predict sales based on advertisement budgets using a multiple linear regression model.

💼 Features:
- TV
- Radio
- Newspaper

Made with ❤️ using Gradio and statsmodels.
"""
```

```python
demo = gr.Interface(
    fn=predict_sales,
    inputs=[
        gr.Slider(0, 300, value=100, label="📺 TV Budget (in $1000s)"),
        gr.Slider(0, 100, value=25, label="📻 Radio Budget (in $1000s)"),
        gr.Slider(0, 120, value=20, label="📰 Newspaper Budget (in $1000s)")
    ],
    outputs=gr.Textbox(label="Predicted Sales"),
    title=title,
    description=description,
    theme=gr.themes.Soft(),
    allow_flagging="never"
)

demo.launch(share=True)
```
